```python
import pandas as pd

df = pd.read_csv('metadata.csv', dtype={'id': 'string'})

# check: 20 < items < 50

print(f'the number of items: {len(df)}')
assert 20 <= len(df) <= 50, 'the number of items does not meet the requirement of 20-50'
```

```python
# check: [id] not duplicated

assert df['id'].is_unique, 'error: the [id] column has duplicate values'
print('pass, all [id] is unique')
```

```python
# check: missing value

print(df.isna().sum())
```
